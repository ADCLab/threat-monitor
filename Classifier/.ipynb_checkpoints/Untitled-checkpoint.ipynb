```python
import json
import re

def parse_block(block):
    # Parse a block of lines representing one comment
    result = {}
    # Parse the comment number from the first line (e.g., "Comment #8:")
    match = re.match(r"Comment #(\d+):", block[0])
    if match:
        result["comment"] = int(match.group(1))
    # Iterate over the rest of the lines in the block
    for line in block[1:]:
        # Match key-value pairs separated by a colon
        match = re.match(r"([^:]+):\s*(.*)", line)
        if match:
            # Normalize the key by stripping extra spaces, converting to lowercase, and replacing spaces with underscores
            key = match.group(1).strip().lower().replace(" ", "_")
            value = match.group(2).strip()
            # Convert "Rating" and "Tokens Used" fields to integers if the value is numeric,
            # and set them to None if the value is "N/A"
            if key == "rating":
                if value.isdigit():
                    value = int(value)
                elif value.upper() == "N/A":
                    value = None
            elif key == "tokens_used":
                if value.isdigit():
                    value = int(value)
                elif value.upper() == "N/A":
                    value = None
            result[key] = value
    return result

def main():
    input_filename = "output_06.txt"
    output_filename = "output_06.jsonl"
    
    # Read the input file and split it into lines
    with open(input_filename, "r", encoding="utf-8") as f:
        lines = f.read().splitlines()
    
    # Group lines into blocks (each block represents one comment)
    blocks = []
    current_block = []
    for line in lines:
        # An empty line indicates the end of a block
        if line.strip() == "":
            if current_block:
                blocks.append(current_block)
                current_block = []
        else:
            current_block.append(line)
    if current_block:
        blocks.append(current_block)
    
    # Open the output file and write each block as a JSON object on a new line
    with open(output_filename, "w", encoding="utf-8") as f:
        for block in blocks:
            parsed = parse_block(block)
            f.write(json.dumps(parsed) + "\n")

main()
```
